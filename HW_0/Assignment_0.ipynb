```python
# linear algebra
print("========== BLOCK 6: CORE LINEAR ALGEBRA ==========")

A = np.array([
    [4., 1., 2.],
    [1., 3., 0.],
    [2., 0., 5.]
])

B = np.array([
    [2., 0., 1.],
    [0., 2., 0.],
    [1., 0., 3.]
])

rhs = np.array([1., 2., 3.])

show("A", A)

show(
    "linalg.inv(A)",
    linalg.inv(A)
)

show(
    "linalg.pinv(A)",
    linalg.pinv(A)
)

show(
    "np.linalg.matrix_rank(A)",
    np.linalg.matrix_rank(A)
)

show(
    "linalg.solve(A, rhs)",
    linalg.solve(A, rhs)
)

# Least squares
A_rect = np.array([
    [1., 1.],
    [1., 2.],
    [1., 3.]
])

b_rect = np.array([1., 2., 2.])

ls_result = linalg.lstsq(A_rect, b_rect)

show(
    "linalg.lstsq(A_rect, b_rect)",
    ls_result
)

# SVD
U, S, Vh = linalg.svd(A)

show("SVD U", U)
show("SVD S", S)
show("SVD Vh", Vh)

# Cholesky
show(
    "linalg.cholesky(A)",
    linalg.cholesky(A)
)

# Eigenvalues
eigvals, eigvecs = linalg.eig(A)

show("eig(A) eigenvalues", eigvals)
show("eig(A) eigenvectors", eigvecs)

# Generalized eigenvalue
eigvals2, eigvecs2 = linalg.eig(A, B)

show("eig(A,B) eigenvalues", eigvals2)
show("eig(A,B) eigenvectors", eigvecs2)

# QR
Q, R = linalg.qr(A)

show("QR Q", Q)
show("QR R", R)

# LU
P, L, U = linalg.lu(A)

show("LU P", P)
show("LU L", L)
show("LU U", U)

show(
    "Check A == P @ L @ U",
    np.allclose(A, P @ L @ U)
)

# Conjugate gradient
x_cg, info = cg(A, rhs)

show("cg solution", x_cg)
show("cg info", info)
```

```python
# Task 4
import numpy as np
import matplotlib.pyplot as plt

# Time values
t = np.linspace(0, 10, 500)

# Damped oscillation
y = np.exp(-0.25 * t) * np.cos(2 * np.pi * t)

# Create the plot
plt.figure(figsize=(9, 5))

plt.plot(t, y, linewidth=2)

# Add labels and title
plt.title("Damped Oscillation")
plt.xlabel("Time")
plt.ylabel("Amplitude")

# Add reference line
plt.axhline(0, linewidth=0.8)

# Add grid
plt.grid(True, alpha=0.3)

plt.show()
```
